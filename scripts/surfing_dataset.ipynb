```python
import pandas as pd
```

```python
constructors = pd.read_csv('../dataset/constructors.csv')
drivers = pd.read_csv('../dataset/drivers.csv')
qualifyings = pd.read_csv('../dataset/qualifying.csv')
circuits = pd.read_csv('../dataset/circuits.csv')
races = pd.read_csv('../dataset/races.csv')

constructors
```

```python
df_merge_races_circuits_qualifyings = pd.merge(races, circuits, on='circuitId')
df_merge_races_circuits_qualifyings = pd.merge(df_merge_races_circuits_qualifyings, qualifyings, on='raceId')
df_merge_races_circuits_qualifyings
```

```python
df_merge = pd.merge(qualifyings, drivers, on='driverId')
df_merge
```

```python
# Search the driver with the most pole positions
pole_positions = df_merge[df_merge['position'] == 1]
pole_counts = pole_positions['driverId'].value_counts()
pole_counts
top_driver_id = pole_counts.idxmax()
top_driver = drivers[drivers['driverId'] == top_driver_id]
top_driver
```

```python
# Search the driver with the most pole positions in every circuit
merge = pd.merge(df_merge, df_merge_races_circuits_qualifyings, on="raceId")
merge
merge_pole_positions = merge[merge['position_x'] == 1]
merge_pole_counts = merge_pole_positions.groupby('circuitId_x')['driverId_x'].value_counts()
merge_pole_counts
top_drivers_per_circuit = merge_pole_counts.groupby(level=0).idxmax()
top_drivers_per_circuit_info = []
for (circuitId, driverId) in top_drivers_per_circuit:
    circuit_name = circuits[circuits['circuitId'] == circuitId]['name'].values[0]
    driver_info = drivers[drivers['driverId'] == driverId][['forename', 'surname']].values[0]
    top_drivers_per_circuit_info.append((circuit_name, driver_info[0], driver_info[1]))
top_drivers_per_circuit_info
```
